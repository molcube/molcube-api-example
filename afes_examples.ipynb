# AMBER TI RB

```python
afes = molcube.create_afes_project()
```

```python
#
# Step 1
#
afes.create_project(
    title='test_amberti_rb',
    reference_project_id='bc2060b8-0bdd-4f3f-97c8-43a83a94e7c8'
)
```

```python
ref_ligands = afes.check_reference_ligands()
```

```python
afes.read_ligand_modeler_project('72cd412f-34f2-435c-9483-f7bbf339561c')
```

```python
selected_ligands = {
    LigandForceField.GAFF_FAST: ['1a.sdf', '1b.sdf', '1c.sdf'], # or ['1a', '1b', '1c']
}

#
# Step 2
#
afes.generate_morph_set(
    ligand_ff=[LigandForceField.GAFF_FAST],
    selected_ligands=selected_ligands,
    generate_morphset_type=GenerateMorphsetType.CLOSED,
)
```

```python
lig_ff = LigandForceField.GAFF_FAST
morph_set = {
    'start': '1c.sdf',
    'end': '1a',
}
afes.delete_morph_set(lig_ff=lig_ff, morph_set=morph_set)
```

```python
lig_ff = LigandForceField.GAFF_FAST
morph_set = {
    'start': '1c.sdf',
    'end': '1a',
}
afes.add_morph_set(lig_ff=lig_ff, morph_set=morph_set)
```

```python
lig_ff = LigandForceField.GAFF_FAST
morph_set = {
    'start': 3,
    'end': 1,
}
afes.delete_morph_set(lig_ff=lig_ff, morph_set=morph_set)
```

```python
lig_ff = LigandForceField.GAFF_FAST
morph_set = {
    'start': 3,
    'end': 1,
}
afes.add_morph_set(lig_ff=lig_ff, morph_set=morph_set)
```

```python
afes.get_similarity_score(lig_ff=LigandForceField.GAFF_FAST)
```

```python
afes.check_ligand_2d_structure(file_name_1='1a.sdf', file_name_2='1c.sdf')
```

```python
afes.processing_ligand()
```

```python
afes.get_core_indices(LigandForceField.GAFF_FAST, set_num=1)
```

```python
afes.delete_core_indices(
    lig_ff=LigandForceField.GAFF_FAST,
    set_num=1,
    ligand_name='1a', indices=[14])
```

```python
afes.add_core_indices(
    lig_ff=LigandForceField.GAFF_FAST,
    set_num=1,
    ligand_name='1a', indices=[14])
```

```python
afes.generate_system()
```

```python
afes.download_project()
```
